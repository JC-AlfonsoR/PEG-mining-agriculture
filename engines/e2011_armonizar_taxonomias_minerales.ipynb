# Poligonos de zonas con potencial mineral

## Cargar datos

````python
potencialMineral_g1 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G1_metales_preciosos/75f31543-bd1b-43bc-942f-99932ea7eba92020412-1-1srbe6r.4f47.shp')
potencialMineral_g2 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G2_metales_basicos/Recursos_Minerales_Grupo2__Metales_Básicos.shp')
potencialMineral_g3 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G3_metales_ind_acero/Zonas_Potenciales_Integradas_Recursos_Minerales_Grupo_3_%3A_Metales_de_la_industria_del_acero.shp')
potencialMineral_g4 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G4_metales_especiales/Zonas_Potenciales_Integradas_Recursos_Minerales_Grupo_4_%3A_Metales_Especiales.shp')
potencialMineral_g5 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G5_minerales_industriales/d5ad3fe1-0afc-4b18-b732-d0efb988f4b1202041-1-1eas1dw.4z55.shp')
potencialMineral_g6 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G6_minerales_energeticos/xn--Zonas_Potenciales_Integradas_Recursos_Minerales_Grupo_6__Minerales_Enrgeticos-8zg.shp')
potencialMineral_g7 = gpd.read_file(DATA/'raw/aba_SGC_zonas_potenciales_minerales/G7_materiales_construccion/b13056a7-dc5a-4eb5-9d8e-947fff89f2762020412-1-1qufl83.jrss.shp')

# Agrupar en lista para facilitar la exploración
lista_dfs = [potencialMineral_g1,
                    potencialMineral_g2,
                    potencialMineral_g3,
                    potencialMineral_g4,
                    potencialMineral_g5,
                    potencialMineral_g6,
                    potencialMineral_g7]
````

## Explorar taxonomía
- En estas bases de datos, las columnas que definien las categorías de las zonas de potencial son ```GR_MINERAL``` y ```SUBGRUPO_M```
- Voy a contar cuantos polígonos hay en cada una de esas categorías

````python
# Explorar las columnas disponibles en cada DF
columnas = pd.DataFrame([df.columns.to_list() for df in lista_dfs])
display(columnas)

# Como los nomobres de las columnas de interes coinciden, puedo concatenar los datos
potencialMineral = pd.concat(lista_dfs)
display(potencialMineral.info())
````

````python
# Contar el numero de registros de cada categoria de mierales
categorias_potencialMineral = potencialMineral.groupby(['SUBGRUPO_M', 'ELEM_PRINC'])['geometry'].count().to_frame().reset_index().sort_values(by='geometry', ascending=False)
display(categorias_potencialMineral.head(3))
````

# Mineria Ilegal

````python
# cargar datos 
SR2021Minerales = pd.read_stata(DATA/"raw/ada_SR2021/panel_prillegalMi.dta")

# explorar datos
display(SR2021Minerales.info())
display(SR2021Minerales.head(3))
````

````python
# La columna relevante para la clasificación es "mineral"
# Contar los registros por las valores de esa columna
display(
    SR2021Minerales.groupby('mineral')['codmpio'].count().reset_index().sort_values(by='codmpio', ascending=False))
````

````python
# Como la base de datos esta codificada, fue necesario encontrar la correspondencia con los minerales
# El diccionario SR2021_mineral_code de /config/utils_for_SR2021.py contiene la correspondencia y la referencia
# que permitió definirlo
display(SR2021_mineral_code)
````

````python
# Crear columna con el nombre del mineral decodificado
SR2021Minerales['mineral_decodificado'] = SR2021Minerales['mineral'].map(SR2021_mineral_code)
````

````python
# Encontrar los categorias de la base de datos
categorias_SR2021Minerales = SR2021Minerales.groupby(['mineral','mineral_decodificado'])['codmpio'].count().reset_index().sort_values(by='codmpio', ascending=False)
display(categorias_SR2021Minerales)
````
