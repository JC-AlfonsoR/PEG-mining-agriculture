# Descargar datos
Los siguientes datos están disponibles para descargar en múltiples archivos.

Esta sección se encarga de descargarlos y organizarlos:
- [Precipitación mensual - CHIRPS 3](https://www.chc.ucsb.edu/data/chirps3)
- [Temperatura max y mine mensual - CHIRTS-era5](https://www.chc.ucsb.edu/data/chirts-era5)

## CHIRPS - Precipitación

```python
# Carpeta destino
carpeta_precipitacion = DATA/"raw/aca_CHIRPS_precipitacion"


# url base precipitacion
url_base_precipitacion = "https://data.chc.ucsb.edu/products/CHIRPS/v3.0/monthly/latam/tifs"

# Configurar reintentos cuanda haya problemas de conexión o errores del servidor
reintentos = Retry(
    total=5,
    backoff_factor=2,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"],
)

# Lista para registrar descargas fallidas
fallidos = []
```

```python
with requests.Session() as sesion:
    sesion.mount("https://", HTTPAdapter(max_retries=reintentos))

    for anno in tqdm(range(1981, 2027)):
        ultimo_mes = 8 if anno == 2026 else 12

        for mes in range(1, ultimo_mes + 1):
            nombre = f"chirps-v3.0.{anno}.{mes:02d}.tif"
            destino = carpeta_precipitacion / nombre
            temporal = carpeta_precipitacion / f"{nombre}.part"

            # Omitir archivos que ya están descargados
            if destino.exists() and destino.stat().st_size > 0:
                print(f"Ya existe: {nombre}")
                continue

            url = f"{url_base_precipitacion}/{nombre}"

            try:
                print(f"Descargando: {nombre}", flush=True)

                with sesion.get(
                    url, stream=True, timeout=(30, 180)
                ) as respuesta:
                    respuesta.raise_for_status()

                    # Descargar por bloques para limitar el uso de memoria
                    with temporal.open("wb") as archivo:
                        for bloque in respuesta.iter_content(
                            chunk_size=1024 * 1024
                        ):
                            if bloque:
                                archivo.write(bloque)

                if temporal.stat().st_size == 0:
                    raise ValueError("El archivo descargado está vacío")

                # Solo asignar el nombre definitivo al completar la descarga
                temporal.replace(destino)

            except (requests.RequestException, OSError, ValueError) as error:
                temporal.unlink(missing_ok=True)
                fallidos.append(nombre)
                print(f"ERROR: {nombre} — {error}")

print(f"\nProceso terminado. Archivos fallidos: {len(fallidos)}")

for nombre in fallidos:
    print(nombre)
```

## CHIRTS - Temperatura

```python
# Carpeta destino
carpeta_temperatura = DATA/"raw/acb_CIHRTS_temperatura"


# url base temperatura
urls_temperatura = {
    "Tmax": "https://data.chc.ucsb.edu/experimental/CHIRTS-ERA5/tmax/tifs/monthly",
    "Tmin": "https://data.chc.ucsb.edu/experimental/CHIRTS-ERA5/tmin/tifs/monthly",
}

"https://data.chc.ucsb.edu/experimental/CHIRTS-ERA5/tmax/tifs/monthly/CHIRTS-ERA5.monthly_Tmax.2024.08.tif"

# Configurar reintentos
reintentos = Retry(
    total=5,
    backoff_factor=2,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"],
)


# Lista para registrar descargas fallidas
fallidos = []

with requests.Session() as sesion:
    sesion.mount("https://", HTTPAdapter(max_retries=reintentos))

    for anno in tqdm(range(1981, 2027)):
        ultimo_mes = 8 if anno == 2026 else 12

        for mes in range(1, ultimo_mes + 1):

            for variable, url_base in urls_temperatura.items():
                nombre = (
                    f"CHIRTS-ERA5.monthly_{variable}.{anno}.{mes:02d}.tif"
                )

                destino = carpeta_temperatura / nombre
                temporal = carpeta_temperatura / f"{nombre}.part"
                url = f"{url_base}/{nombre}"

                # Comprobar cada archivo de manera independiente
                if destino.exists() and destino.stat().st_size > 0:
                    tqdm.write(f"Ya existe: {nombre}")
                    continue

                try:
                    tqdm.write(f"Descargando: {nombre}")

                    with sesion.get(
                        url, stream=True, timeout=(30, 180)
                    ) as respuesta:
                        respuesta.raise_for_status()

                        with temporal.open("wb") as archivo:
                            for bloque in respuesta.iter_content(
                                chunk_size=1024 * 1024
                            ):
                                if bloque:
                                    archivo.write(bloque)

                    if temporal.stat().st_size == 0:
                        raise ValueError("El archivo descargado está vacío")

                    temporal.replace(destino)

                except (
                    requests.RequestException,
                    OSError,
                    ValueError,
                ) as error:
                    temporal.unlink(missing_ok=True)
                    fallidos.append(nombre)
                    tqdm.write(f"ERROR: {nombre} — {error}")

print(f"\nProceso terminado. Archivos fallidos: {len(fallidos)}")

for nombre in fallidos:
    print(nombre)
```
